Imporrting dependencies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

Importing house price dataset

```python
house_price_dataframe = pd.read_csv("/content/BostonHousing.csv")
```

```python
print(house_price_dataframe)
```

```python
#print first 5 rows of dataframe
house_price_dataframe.head()
```

```python
#checking the number of rows and columns in the dataframe
house_price_dataframe.shape
```

```python
#check for missing values
house_price_dataframe.isnull().sum()
```

```python
print(house_price_dataframe.isnull().sum())
```

```python
house_price_dataframe['rm']=house_price_dataframe['rm'].fillna(house_price_dataframe['rm'].median())
```

```python
print(house_price_dataframe.isnull().sum())
```

```python
# statistical measures of the dataset
house_price_dataframe.describe()
```

Understanding the correlation btw various features in the dataset



1.   Positive correlation
2.   Negative correlation



```python
correlation = house_price_dataframe.corr()
```



*   cbar -->  do we need color bar or not.
*   square --> all the values should be in square
*   fmt  --> upto which float value
*   annot --> values and feature names in the chart
*   annot_kws -->  size of text and feature name
*   cmap -->  color of heat map



```python
#constructing the heat map to understand the correlation
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')

```

Splitting the data and target

```python
X=house_price_dataframe.drop(['medv'],axis=1)
Y=house_price_dataframe['medv']
```

```python
print(X)
```

```python
print(Y)
```

Splittng data into training and testing data

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y,test_size=0.2,random_state = 2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Model Training

XGBoost Regressor

```python
# loading the model
model = XGBRegressor()
```

```python
# trainig the model with X_train
model.fit(X_train,Y_train)
```

Evaluation

Prediction on training data

```python
# accuracy for prediction on training data
training_data_prediction = model.predict(X_train)
print(training_data_prediction)
```

```python
# R square error
score_1 = metrics.r2_score(Y_train,training_data_prediction)

# Mean absolute error
score_2 = metrics.mean_absolute_error(Y_train,training_data_prediction)

print("R square error ",score_1)      # it should be closer to zero

print("Mean absolute error: ",score_2)
```

Visualizing the actual Prices and predicted price

```python
plt.scatter(Y_train,training_data_prediction)
plt.xlabel("Actutal Price")
plt.ylabel("Predicted Price")
plt.title("Actual Price vs Predicted Price")
plt.show()
```

Prediction on test data


```python
# accuracy for prediction on testing data
testing_data_prediction = model.predict(X_test)
print(testing_data_prediction)
```

```python
# R square error
score_1 = metrics.r2_score(Y_test,testing_data_prediction)

# Mean absolute error
score_2 = metrics.mean_absolute_error(Y_test,testing_data_prediction)

print("R square error ",score_1)      # it should be closer to zero

print("Mean absolute error: ",score_2)
```
